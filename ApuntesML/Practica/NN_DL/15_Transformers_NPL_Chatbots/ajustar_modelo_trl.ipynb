# Ajustando un modelo utilizando la libreria TRL

```python
from datasets import load_dataset
sft_dataset = load_dataset("tatsu-lab/alpaca", split="train")
print(sft_dataset[1]["text"])
```

Como se puede ver en el log del dataset, se ha preentrenado para que el modelo siga una única instrucción y genere una respuesta útil y coerente. Es un buen punto de partida, pero será útil continuar ajustando el modelo para utilizar un conjunto multiturno para desarrollar la habilidad del modelo de tener conversaciones largas. Además esto hará que el modelo aprenda a generar etiquetas de rol, como "Usuario:" "Asistente:" 

```python
# Funcion para preprocesar las etiquetas de rol y ajustarlas al estandar de Anthropic 
# Cada ejemplo en el dataset contiene un prompt en un campo de "texto", asi como sus componentes
# en campos separados example['instruction'], example['output'] y opcionalmente example['input']
def preprocess(example):
    text = f"Human: {example['instruction']}\n"
    if example['input'] != "":
        text += f"-> {example['input']}\n"
    text += f"\nAssistant: {example['output']}"
    return {"text": text}
    
sft_dataset = sft_dataset.map(preprocess)

print(sft_dataset[1]["text"])
```

```python
from trl import SFTTrainer, SFTConfig
sft_model_dir = "./my_gpt2_sft_alpaca"
training_args = SFTConfig(output_dir=sft_model_dir, max_length=512,
per_device_train_batch_size=4, num_train_epochs=1, save_steps=50,
logging_steps=10, save_total_limit=2, learning_rate=5e-5)

sft_trainer = SFTTrainer("gpt2", train_dataset=sft_dataset, args=training_args)
sft_train_output = sft_trainer.train()
sft_trainer.model.save_pretrained(sft_model_dir)
```

```python
# El conjuntode Anthropic de hh-rlhf es utilizado para sanear las respuestas del chatbot
pref_dataset = load_dataset("Anthropic/hh-rlhf", split="train")
pref_dataset[2].keys()
print(pref_dataset[2]["chosen"].strip())
print(pref_dataset[2]["rejected"].strip())
```

```python
# En este conjunto el prompt esta incluido tanto en las respuestas aceptadas como en las rechazadas. En otros estan
# en su propia sección "prompt". La libreria de trl puede manejar ambos casos, con lo que continuaremos con la segunda fase del
# ajuste utilizando DPO
from trl import DPOConfig, DPOTrainer
from transformers import AutoTokenizer
model_id = "gpt2"
gpt2_tokenizer = AutoTokenizer.from_pretrained(model_id)
dpo_model_dir = "./my_gpt2_sft_alpaca_dpo_hh_rlhf"
training_args = DPOConfig(
output_dir=dpo_model_dir, max_length=512, per_device_train_batch_size=4,
num_train_epochs=1, save_steps=50, logging_steps=10, save_total_limit=2, learning_rate=2e-5)
gpt2_tokenizer.pad_token = gpt2_tokenizer.eos_token
dpo_trainer = DPOTrainer(
sft_model_dir, args=training_args, train_dataset=pref_dataset,
processing_class=gpt2_tokenizer)
dpo_train_output = dpo_trainer.train()
dpo_trainer.model.save_pretrained(dpo_model_dir)
```

Hecho esto, ya se ha cubierto como implementar un transformer desde 0, preentrenarlo con NTP (teniendo los recursos necesarios) y ajustarlo utilizando SFT y DPO para convertirlo en un chatbot.
